```python
import pandas as pd
```

```python
df = pd.read_parquet('./data/calibration_viewser_df_raw.parquet')
```

```python
df
```

```python
df.describe()
```

```python
missing_count = df.isna().sum().rename("missing_count")
missing_pct = (df.isna().mean() * 100).rename("missing_pct")

missing_summary = pd.concat([missing_count, missing_pct], axis=1)

missing_summary
```

```python
df_selected = df.loc[df.index.get_level_values(0).isin(range(385, 540))]
```

```python
df_selected
```

```python
missing_count = df_selected.isna().sum().rename("selected_missing_count")
missing_pct = (df_selected.isna().mean() * 100).rename("selected_missing_pct")

selected_missing_summary = pd.concat([missing_count, missing_pct], axis=1)

selected_missing_summary
```

```python
pd.merge(missing_summary, selected_missing_summary, left_index=True, right_index=True)
```

```python
train_start, train_end = 121, 444
time_idx = df.index.get_level_values(0)
train = df.loc[(time_idx >= train_start) & (time_idx <= train_end)]

nan_count_by_country_in_train = train.isna().groupby(level="country_id").sum()

n_rows_by_country = train.groupby(level="country_id").size()

for col in nan_count_by_country_in_train.columns:
  mask_all_nan = nan_count_by_country_in_train[col].eq(n_rows_by_country)
  country_ids = nan_count_by_country_in_train.index[mask_all_nan].tolist()
  print(f"{col}: {country_ids}")
```

```python
# Missing percentage per column after excluding fully-missing countries
level = "country_id"

nan_share_by_country = (
    df.isna()
      .groupby(level=level)
      .mean()
)

rows = []
for col in df.columns:
    fully_missing_countries = nan_share_by_country.index[nan_share_by_country[col] == 1.0]
    filtered = df.loc[~df.index.get_level_values(level).isin(fully_missing_countries), col]

    rows.append({
        "column": col,
        "remaining_missing_pct_after_exclusion": filtered.isna().mean() * 100,
    })

missing_after_exclusion_summary = (
    pd.DataFrame(rows)
      .set_index("column")
)

missing_after_exclusion_summary
```

```python
missing_summary.merge(
  missing_after_exclusion_summary,
  left_index=True,
  right_index=True
)
```
